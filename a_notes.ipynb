# Program design
This document represents our first approach. It will split and store the tasks.

## Requirements
- write a simple command-line Tic-Tac-Toe game in Python that two players can play against each other

## Interface
- start dialoge (welcome players)
- a 3x3 board
- figures (e.g. 'x' and 'o')
- mark used fields on the board

## Rules
- store field state (free/used)
- win condition (3 adjacent fields used by same figure)
- start over or end programm

## Input
- user figure selection
- user figure placing on field

## Flow
- store users (name, figure)
- store action: fields state
- check win condition

## Output
- update fields state
- notify at win condition
