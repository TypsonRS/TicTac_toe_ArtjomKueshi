# Program design
This document represents our first approach. It will split and store the tasks.

## Requirements
- write a simple command-line Tic-Tac-Toe game in Python that two players can play against each other

## Interface
- start dialoge
- a size selection and minimum 3x3 board
- user names
- user symbols (e.g. 'x' and 'o')
- mark used fields on the board

## Rules
- store field state (free/used)
- win condition (3 adjacent fields used by same figure)
- start over or end programm

## Input
- user figure placing on field

## Flow
- store users (name, figure)
- store action: fields state
- check win condition

## Output
- update fields state
- notify at win condition

```python
# early version of board_square
def board_square():
    """
    Takes initial user input for board size, finds the next perfect square
    """
    print("Welcome to the board generator!")
    x = int(input('Please enter an integer or hit <Enter> to use a standard of 9:') or 9)
    if x < 7:
        x = 9
    y = 1
    while abs(x - (y ** 2)) > abs(x - ((y + 1) ** 2)):          # at some time I want to think that through: it gives the closest perfect square
        y += 1
    board_square = y ** 2
    very_nice()
    print("Board size:", board_square)
    return y
board_square()
```

```python
# early version of board generator with empty fields
def board_generator():
    """
        Takes the board size and creates the board
    """
    y = board_square()
    board_size = y ** 2
    board_fields = list(range(1,board_size+1))
    board_dictionary = dict((el,' ') for el in board_fields)     # todo: actually study comprehensions
    return board_dictionary
board_generator()
```

```python
# early version of board printer for analysis
def board_printer():
    """
        Prints the board with values
    """
    board_dictionary = board_generator()
    y = board_square()
    print("Here is the board:")
    print('+---' * y + '+')
    for row in range(y):
        row_values = [board_dictionary[row * y + col + 1] for col in range(y)]  # todo: think that through
        print("| " + " | ".join(row_values) + " |")              # todo: join
        if row < y - 1:
            print('+---' * y + '+')
    print('+---' * y + '+')
    return board_printer()
```
