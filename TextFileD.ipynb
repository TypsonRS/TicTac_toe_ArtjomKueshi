```python
# Function for displaying the board
def display_board(board):
    print()
    print(f" {board[0]} | {board[1]} | {board[2]} ")
    print("---+---+---")
    print(f" {board[3]} | {board[4]} | {board[5]} ")
    print("---+---+---")
    print(f" {board[6]} | {board[7]} | {board[8]} ")
    print()

```

```python
# Function for choosing a player
def choose_player():
    player = input("Choose your symbol (X or O): ").upper()

    while player not in ["X", "O"]:
        print("Please choose X or O.")
        player = input("Choose your symbol (X or O): ").upper()

    return player
```

```python
# Function for displaying the available positions
def display_positions():
    print("Board positions:")
    print()
    print(" 1 | 2 | 3 ")
    print("---+---+---")
    print(" 4 | 5 | 6 ")
    print("---+---+---")
    print(" 7 | 8 | 9 ")
    print()
```

```python
# Function for checking if a player has won
def check_winner(board, player):
    winning_combinations = [
        [0, 1, 2],
        [3, 4, 5],
        [6, 7, 8],
        [0, 3, 6],
        [1, 4, 7],
        [2, 5, 8],
        [0, 4, 8],
        [2, 4, 6]
    ]

    for combination in winning_combinations:
        if all(board[position] == player for position in combination):
            return True

    return False
```

```python
# Function for checking if the board is full
def board_full(board):
    return " " not in board
```

```python
# Function for making a move
def make_move(board, player):
    while True:
        try:
            position = int(input(f"Player {player}, choose a position (1-9): "))

            if position < 1 or position > 9:
                print("Please choose a number between 1 and 9.")
                continue

            index = position - 1

            if board[index] != " ":
                print("That position is already taken.")
                continue

            board[index] = player
            break

        except ValueError:
            print("Please enter a number.")
```

```python
# Tic-tac-toe game
if __name__ == "__main__":

    # Start a new round of Tic-Tac-Toe
    print("Welcome to a new round of Tic-Tac-Toe!")

    # Create an empty board
    board = [" "] * 9

    # Show the positions
    display_positions()

    # Choose the first player
    current_player = choose_player()

    # Game loop
    while True:

        # Display the current board
        display_board(board)

        # Player makes a move
        make_move(board, current_player)

        # Check if the player has won
        if check_winner(board, current_player):
            display_board(board)
            print(f"Congratulations! Player {current_player} wins!")
            break

        # Check if the board is full
        if board_full(board):
            display_board(board)
            print("It's a draw!")
            break

        # Change player
        if current_player == "X":
            current_player = "O"
        else:
            current_player = "X"
```
